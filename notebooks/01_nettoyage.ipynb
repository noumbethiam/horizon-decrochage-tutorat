```python
df_insc["statut_final"] = df_insc["statut_final"].str.strip().str.lower()

correspondance = {
    "inscrit": "inscrit",
    "abandon": "abandon", "abandonné": "abandon",
    "transfert": "transfert", "transfere": "transfert", "transféré": "transfert",
}
df_insc["statut_final"] = df_insc["statut_final"].map(correspondance)

print(df_insc["statut_final"].value_counts(dropna=False))

sans_statut = df_insc[df_insc["statut_final"].isna()]
print("sans statut mais avec date_fin :", sans_statut["date_fin"].notna().sum())
display(sans_statut.head(10))
```

```python
a_date_fin = df_insc["date_fin"].notna()
statut = df_insc["statut_final"].fillna("(vide)")

print(pd.crosstab(statut, a_date_fin))
```

```python
print(df_insc["date_debut"].value_counts().sort_index())
print(df_insc["date_fin"].dropna().sample(15, random_state=1))
```

```python
mois = {"janv.": "01", "févr.": "02", "mars": "03", "avr.": "04", "mai": "05", "juin": "06",
        "juil.": "07", "août": "08", "sept.": "09", "oct.": "10", "nov.": "11", "déc.": "12"}
formats = ["%Y-%m-%d", "%Y/%m/%d", "%d/%m/%Y", "%d %m %Y"]

def convertir_date(texte):
    if pd.isna(texte):
        return pd.NaT
    t = str(texte).strip().lower()
    for nom, num in mois.items():
        t = t.replace(nom, num)
    for f in formats:
        try:
            return pd.to_datetime(t, format=f)
        except ValueError:
            pass
    return pd.NaT

df_insc["date_debut"] = df_insc["date_debut"].apply(convertir_date)
df_insc["date_fin"] = df_insc["date_fin"].apply(convertir_date)

print(df_insc[["date_debut", "date_fin"]].dtypes)
print("date_debut non lues :", df_insc["date_debut"].isna().sum())
print("date_fin vides :", df_insc["date_fin"].isna().sum())
print(df_insc["date_debut"].min(), df_insc["date_debut"].max())
print(df_insc["date_fin"].min(), df_insc["date_fin"].max())
```

```python
rentrees = pd.to_datetime(["2024-10-07", "2025-10-06"])
atypiques = ~df_insc["date_debut"].isin(rentrees)

duree = (df_insc["date_fin"] - df_insc["date_debut"]).dt.days / 7

print("début atypique :", atypiques.sum())
print("fin avant début :", (duree < 0).sum())
print("fin après 30 semaines :", (duree > 30).sum())
print(duree.describe())

display(df_insc[atypiques])
```

```python
inverses = df_insc["date_fin"] < df_insc["date_debut"]

debut_corrige = df_insc.loc[inverses, "date_fin"].copy()
fin_corrigee = df_insc.loc[inverses, "date_debut"].copy()
df_insc.loc[inverses, "date_debut"] = debut_corrige
df_insc.loc[inverses, "date_fin"] = fin_corrigee

rentrees = pd.to_datetime(["2024-10-07", "2025-10-06"])
duree = (df_insc["date_fin"] - df_insc["date_debut"]).dt.days / 7

print("lignes échangées :", inverses.sum())
print("début atypique :", (~df_insc["date_debut"].isin(rentrees)).sum())
print("fin avant début :", (duree < 0).sum())
print("fin après 30 semaines :", (duree > 30).sum())

display(df_insc[duree > 30])
```

```python
trop_tard = ((df_insc["date_fin"] - df_insc["date_debut"]).dt.days / 7) > 30
df_insc.loc[trop_tard, "date_fin"] = pd.NaT

print("dates de fin mises à vide :", trop_tard.sum())
print("date_fin vides :", df_insc["date_fin"].isna().sum())
print(df_insc.loc[trop_tard, "statut_final"].value_counts())
```

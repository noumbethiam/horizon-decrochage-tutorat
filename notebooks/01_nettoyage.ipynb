```python
mois = {"janv.": "01", "févr.": "02", "mars": "03", "avr.": "04", "mai": "05", "juin": "06",
        "juil.": "07", "août": "08", "sept.": "09", "oct.": "10", "nov.": "11", "déc.": "12"}
formats = ["%Y-%m-%d", "%Y/%m/%d", "%d/%m/%Y", "%d %m %Y"]

def convertir_date(texte):
    if pd.isna(texte):
        return pd.NaT
    t = str(texte).strip().lower()
    for nom, num in mois.items():
        t = t.replace(nom, num)
    for f in formats:
        try:
            return pd.to_datetime(t, format=f)
        except ValueError:
            pass
    return pd.NaT

df_insc["date_debut"] = df_insc["date_debut"].apply(convertir_date)
df_insc["date_fin"] = df_insc["date_fin"].apply(convertir_date)

print(df_insc[["date_debut", "date_fin"]].dtypes)
print("date_debut non lues :", df_insc["date_debut"].isna().sum())
print("date_fin vides :", df_insc["date_fin"].isna().sum())
print(df_insc["date_debut"].min(), df_insc["date_debut"].max())
print(df_insc["date_fin"].min(), df_insc["date_fin"].max())
```

```python
act = act.sort_values(["id_etudiant", "semaine"])
act["inactif"] = (act["nb_connexions_lms"] == 0) & (act["devoirs_rendus"] == 0)

def premiere_inactivite(semaines, inactifs, k):
    compteur = 0
    for sem, inact_ in zip(semaines, inactifs):
        compteur = compteur + 1 if inact_ else 0
        if compteur == k:
            return sem - k + 1
    return None

resultats = {}
for etu, g in act.groupby("id_etudiant"):
    sem, ina = g["semaine"].values, g["inactif"].values
    resultats[etu] = {k: premiere_inactivite(sem, ina, k) for k in (3, 4, 5)}

inact = pd.DataFrame.from_dict(resultats, orient="index")
inact.columns = ["inact3", "inact4", "inact5"]

d = d.set_index("id_etudiant").join(inact)
print(d.groupby("statut_final")[["inact3", "inact4", "inact5"]].count())
print(pd.crosstab(d["statut_final"], d["inact4"].notna()))

avec = d[(d["statut_final"] == "abandon") & d["sem_fin"].notna() & d["inact4"].notna()]
print((avec["sem_fin"] - avec["inact4"]).value_counts().sort_index().head(10))
```

```python
etu = pd.read_csv("../data/processed/etudiants.csv")
tut = pd.read_csv("../data/processed/tutorat.csv")

base = etu.set_index("id_etudiant")
base = base.join(par_etu[["derniere"]]).join(inact)
base = base.join(insc.set_index("id_etudiant")[["annee_universitaire", "statut_final", "sem_fin"]])
base["statut_final"] = base["statut_final"].fillna("(sans inscription)")

def construire(base, k):
    col = f"inact{k}"
    declare = base["statut_final"] == "abandon"
    transfert = base["statut_final"] == "transfert"
    regle = base[col].notna() & ~declare & ~transfert
    T = base["derniere"].astype(float)
    E = pd.Series(0, index=base.index)
    T = T.where(~declare, base["sem_fin"].fillna(base["derniere"]))
    E = E.where(~declare, 1)
    T = T.where(~transfert, base["sem_fin"].fillna(base["derniere"]))
    T = T.where(~regle, base[col])
    E = E.where(~regle, 1)
    return T.astype(int), E.astype(int)

for k in (3, 4, 5):
    T, E = construire(base, k)
    suffixe = "" if k == 4 else f"_k{k}"
    base[f"T{suffixe}"] = T
    base[f"E{suffixe}"] = E

tut2 = tut.set_index("id_etudiant")[["semaine_debut", "type_tuteur", "motif_orientation"]]
tut2.columns = ["semaine_debut_tutorat", "type_tuteur", "motif_orientation"]
base = base.join(tut2)
base["tutore"] = base.index.isin(tut["id_etudiant"]).astype(int)
base = base.drop(columns=["derniere", "sem_fin", "inact3", "inact4", "inact5"])

print(base.shape, base.index.is_unique)
print(pd.crosstab(base["statut_final"], base["E"]))
print("T de", base["T"].min(), "à", base["T"].max())
print("taux de censure (E=0) :", round((base["E"] == 0).mean(), 3))
print("événements : k=4 :", base["E"].sum(), "| k=3 :", base["E_k3"].sum(), "| k=5 :", base["E_k5"].sum())
print("tutorés :", base["tutore"].sum())
print("tutorat commencé après T :", (base["semaine_debut_tutorat"] > base["T"]).sum())

base.reset_index().to_csv("../data/processed/base_survie.csv", index=False)
print(pd.read_csv("../data/processed/base_survie.csv").shape)
```

## Interprétation de l'analyse exploratoire (DA-08)

**Censure.** 3560 étudiants sur 5000 (71,2 %) n'ont pas décroché pendant les 30 semaines : ils sont encore inscrits ou ont été transférés. 1440 étudiants (28,8 %) ont décroché. Les étudiants censurés restent dans l'analyse.

**Quand décroche-t-on ?** Les décrochages se concentrent à deux moments : entre les semaines 3 et 8 (jusqu'à 110 décrochages en semaine 6), puis en semaine 16 (123 décrochages, le maximum). Ils deviennent nettement moins nombreux après la semaine 17.

**Profils associés à plus de décrochage.** Avoir un emploi parallèle (43,0 % contre 25,5 %), avoir une mention Passable au bac (36,9 % contre 12,8 % pour Très bien), ne pas être boursier (31,3 % contre 25,2 %). Il n'y a pas de différence nette entre les deux cohortes ni entre les genres. Ces chiffres ne tiennent pas compte du moment du départ : les courbes de Kaplan-Meier les préciseront.

**Tutorat.** Les étudiants tutorés décrochent moins souvent (22,7 % contre 30,8 %), mais cette comparaison est trompeuse : ils avaient de moins bonnes notes au bac, habitent plus loin et sont plus souvent boursiers, et ils devaient encore être inscrits pour commencer le tutorat. On ne peut donc pas conclure à un effet du tutorat à ce stade.

```python
from lifelines import KaplanMeierFitter
from lifelines.statistics import multivariate_logrank_test

base = pd.read_csv("../data/processed/base_survie.csv")

kmf = KaplanMeierFitter()
kmf.fit(base["T"], event_observed=base["E"], label="tous les étudiants")
print("médiane de survie :", kmf.median_survival_time_)
print(kmf.survival_function_at_times([10, 20, 30]).round(3))

fig, ax = plt.subplots(figsize=(7, 4.5))
kmf.plot_survival_function(ax=ax)
ax.set_xlabel("semaines")
ax.set_ylabel("part des étudiants encore là")
ax.set_title("Kaplan-Meier global")
plt.savefig("../docs/figures/km_global.png", dpi=120)
plt.show()

for col in ["boursier", "emploi_parallele", "mention_bac", "filiere"]:
    sous = base.dropna(subset=[col])
    fig, ax = plt.subplots(figsize=(7, 4.5))
    for valeur, g in sous.groupby(col):
        KaplanMeierFitter().fit(g["T"], g["E"], label=str(valeur)).plot_survival_function(ax=ax, ci_show=False)
    p = multivariate_logrank_test(sous["T"], sous[col], sous["E"]).p_value
    ax.set_title(f"Survie selon {col} (log-rank p = {p:.2g})")
    ax.set_xlabel("semaines")
    ax.set_ylabel("part des étudiants encore là")
    plt.savefig(f"../docs/figures/km_{col}.png", dpi=120)
    plt.show()
    print(col, ": p =", p)
```

```python
base = pd.read_csv("../data/processed/base_survie.csv")

# 1. semaines critiques : taux de décrochage parmi les étudiants encore présents
lignes = []
for sem in range(1, 31):
    a_risque = (base["T"] >= sem).sum()
    decroche = ((base["T"] == sem) & (base["E"] == 1)).sum()
    lignes.append({"semaine": sem, "a_risque": a_risque, "decrochages": decroche,
                   "taux": round(decroche / a_risque, 3)})
hazard = pd.DataFrame(lignes)
print(hazard.nlargest(6, "taux"))

# 2. survie à 10, 20 et 30 semaines, par groupe
resume = []
for col in ["boursier", "emploi_parallele", "mention_bac", "filiere"]:
    for valeur, g in base.dropna(subset=[col]).groupby(col):
        s = KaplanMeierFitter().fit(g["T"], g["E"]).survival_function_at_times([10, 20, 30]).values
        resume.append({"variable": col, "groupe": valeur, "etudiants": len(g),
                       "S(10)": round(s[0], 3), "S(20)": round(s[1], 3), "S(30)": round(s[2], 3)})
tableau = pd.DataFrame(resume)
display(tableau)
```

## Interprétation de l'analyse de survie descriptive (SV-01)

**Courbe globale.** 85,3 % des étudiants sont encore inscrits à la semaine 10, 75,2 % à la semaine 20 et 70,9 % à la semaine 30. La médiane de survie n'existe pas : moins de la moitié des étudiants décroche pendant les 30 semaines.

**Semaines critiques.** Le risque de décrocher est le plus élevé en semaine 16 : 3,1 % des étudiants encore présents décrochent cette semaine-là. Viennent ensuite les semaines 6 (2,4 %), 5 (2,3 %), 4 (2,0 %) et 15 (1,9 %). Il y a donc deux périodes sensibles : le début du cursus (semaines 4 à 7) et les semaines 15 et 16.

**Profils associés à un décrochage plus fréquent.** Les tests du log-rank montrent des écarts qui ne s'expliquent pas par le hasard :
- Mention au bac (p < 0,001) : à 30 semaines, 62,9 % des étudiants avec une mention Passable sont encore là, contre 87,1 % avec une mention Très bien.
- Emploi parallèle (p < 0,001) : 56,7 % avec emploi contre 74,2 % sans emploi.
- Statut de boursier (p < 0,001) : 74,5 % des boursiers contre 68,3 % des non-boursiers.
- Filière (p = 0,0023) : l'écart est plus faible, de 68,2 % en Informatique à 74,4 % en Gestion.

Ces résultats décrivent des associations. Ils ne disent pas que l'emploi ou la mention provoque le décrochage, et ils ne tiennent pas compte des autres caractéristiques en même temps : c'est le rôle du modèle de Cox.

```python
mois = {"janv.": "01", "févr.": "02", "mars": "03", "avr.": "04", "mai": "05", "juin": "06",
        "juil.": "07", "août": "08", "sept.": "09", "oct.": "10", "nov.": "11", "déc.": "12"}
formats = ["%Y-%m-%d", "%Y/%m/%d", "%d/%m/%Y", "%d %m %Y"]

def convertir_date(texte):
    if pd.isna(texte):
        return pd.NaT
    t = str(texte).strip().lower()
    for nom, num in mois.items():
        t = t.replace(nom, num)
    for f in formats:
        try:
            return pd.to_datetime(t, format=f)
        except ValueError:
            pass
    return pd.NaT

df_insc["date_debut"] = df_insc["date_debut"].apply(convertir_date)
df_insc["date_fin"] = df_insc["date_fin"].apply(convertir_date)

print(df_insc[["date_debut", "date_fin"]].dtypes)
print("date_debut non lues :", df_insc["date_debut"].isna().sum())
print("date_fin vides :", df_insc["date_fin"].isna().sum())
print(df_insc["date_debut"].min(), df_insc["date_debut"].max())
print(df_insc["date_fin"].min(), df_insc["date_fin"].max())
```

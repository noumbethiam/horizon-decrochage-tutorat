```python
import pandas as pd
df_etudiants = pd.read_csv("../data/raw/etudiants.csv")
print("Fichier chargé :", df_etudiants.shape)
```

```python
print (df_etudiants["genre"].value_counts())
```

```python
df_clean = df_etudiants.copy()

df_clean["genre"] = df_clean["genre"].str.strip().str.lower()

correspondance = {
    "m": "M", "masculin": "M", "homme": "M", "h": "M",
    "f": "F", "femme": "F", "feminin": "F",
}
df_clean["genre"] = df_clean["genre"].map(correspondance)

print(df_clean["genre"].value_counts(dropna=False))
```

```python
print(df_clean["boursier"].value_counts(dropna=False))
```

```python
df_clean["boursier"] = df_clean["boursier"].str.strip().str.lower()

correspondance = {
    "oui": "Oui", "o": "Oui", "yes": "Oui", "1": "Oui",
    "non": "Non", "n": "Non", "no": "Non", "0": "Non",
}
df_clean["boursier"] = df_clean["boursier"].map(correspondance)

print(df_clean["boursier"].value_counts(dropna=False))
```

```python
print(df_clean["emploi_parallele"].value_counts(dropna=False))
```

```python
print(df_clean["filiere"].value_counts(dropna=False).sort_index())
```

```python
df_clean["filiere"] = df_clean["filiere"].str.strip().str.lower()

correspondance = {
    "communication": "Communication", "com": "Communication",
    "droit": "Droit",
    "genie civil": "Génie civil", "génie civil": "Génie civil", "gc": "Génie civil",
    "gestion": "Gestion",
    "informatique": "Informatique", "info": "Informatique",
}
df_clean["filiere"] = df_clean["filiere"].map(correspondance)

print(df_clean["filiere"].value_counts(dropna=False))
```

```python
print(df_clean["mention_bac"].value_counts(dropna=False).sort_index())
```

```python
df_clean["mention_bac"] = df_clean["mention_bac"].str.strip().str.lower()

correspondance = {
    "passable": "Passable", "p": "Passable",
    "assez bien": "Assez bien", "ab": "Assez bien",
    "bien": "Bien", "b": "Bien",
    "très bien": "Très bien", "tres bien": "Très bien", "tb": "Très bien",
}
df_clean["mention_bac"] = df_clean["mention_bac"].map(correspondance)

print(df_clean["mention_bac"].value_counts(dropna=False))
```

```python
print(df_clean["region_origine"].value_counts(dropna=False).sort_index())
```

```python
df_clean["region_origine"] = df_clean["region_origine"].str.strip().str.lower()

correspondance = {
    "dakar": "Dakar",
    "diourbel": "Diourbel",
    "fatick": "Fatick",
    "kaffrine": "Kaffrine",
    "kaolack": "Kaolack",
    "kedougou": "Kédougou", "kédougou": "Kédougou",
    "kolda": "Kolda",
    "louga": "Louga",
    "matam": "Matam",
    "saint-louis": "Saint-Louis", "saint louis": "Saint-Louis",
    "st-louis": "Saint-Louis", "st louis": "Saint-Louis",
    "sedhiou": "Sédhiou", "sédhiou": "Sédhiou",
    "tambacounda": "Tambacounda",
    "thies": "Thiès", "thiès": "Thiès",
    "ziguinchor": "Ziguinchor", "zig": "Ziguinchor",
}
df_clean["region_origine"] = df_clean["region_origine"].map(correspondance)

print(df_clean["region_origine"].value_counts(dropna=False))
```

```python
print(df_clean["age"].value_counts(dropna=False).sort_index())
```

```python
df_clean["age"] = df_clean["age"].str.replace(" ans", "", regex=False)
df_clean["age"] = pd.to_numeric(df_clean["age"])
df_clean["age"] = df_clean["age"].where(df_clean["age"].between(17, 35))

print(df_clean["age"].isna().sum())
print(df_clean["age"].describe())
```

```python
notes = pd.to_numeric(df_clean["note_bac"], errors="coerce")

print(df_clean.loc[notes.isna(), "note_bac"].value_counts(dropna=False))
print(notes.describe())
```

```python
df_clean["note_bac"] = df_clean["note_bac"].str.replace(",", ".", regex=False)
df_clean["note_bac"] = pd.to_numeric(df_clean["note_bac"], errors="coerce")

print(df_clean["note_bac"].isna().sum())

hors_plage = df_clean.loc[~df_clean["note_bac"].between(10, 20), "note_bac"]
print(hors_plage.value_counts().sort_index())
```

```python
df_clean["note_bac"] = df_clean["note_bac"].where(df_clean["note_bac"].between(10, 20))

print(df_clean["note_bac"].isna().sum())
print(df_clean["note_bac"].describe())
```

```python
df_clean = df_clean.rename(columns={"Distance campus (km)": "distance_campus_km"})

distances = pd.to_numeric(df_clean["distance_campus_km"], errors="coerce")

print(df_clean.loc[distances.isna(), "distance_campus_km"].value_counts(dropna=False).head(10))
print(distances.describe())
```

```python
df_clean["distance_campus_km"] = (
    df_clean["distance_campus_km"]
    .str.replace("km", "", regex=False)
    .str.replace(",", ".", regex=False)
    .str.strip()
)
df_clean["distance_campus_km"] = pd.to_numeric(df_clean["distance_campus_km"], errors="coerce")

print(df_clean["distance_campus_km"].isna().sum())
print(df_clean["distance_campus_km"].nsmallest(10))
print(df_clean["distance_campus_km"].nlargest(10))
```

```python
d = df_clean["distance_campus_km"]

print("négatives ou nulles :", (d <= 0).sum())
print("égales à 9999 :", (d == 9999).sum())

normales = d[(d > 0) & (d < 9999)]
print(normales.describe())
print(normales.nlargest(10))
```

```python
d = df_clean["distance_campus_km"]
df_clean["distance_campus_km"] = d.where((d > 0) & (d < 9999))

print(df_clean["distance_campus_km"].isna().sum())
print(df_clean["distance_campus_km"].describe())
```
